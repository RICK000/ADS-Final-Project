```python
total_income = df.loc[df['Income/Expense'] == 'Income', 'Amount'].sum()
total_expense = df.loc[df['Income/Expense'] == 'Expense', 'Amount'].sum()
net_savings = total_income - total_expense

print("Total Income:", total_income)
print("Total Expense:", total_expense)
print("Net Savings:", net_savings)
```

```python
monthly = df.groupby(['Year', 'Month', 'Month_Name', 'Income/Expense'])['Amount'].sum().reset_index()

monthly
```

```python
monthly_summary = df.pivot_table(
    index=['Year', 'Month', 'Month_Name'],
    columns='Income/Expense',
    values='Amount',
    aggfunc='sum',
    fill_value=0
).reset_index()

monthly_summary['Savings'] = (
    monthly_summary.get('Income', 0) -
    monthly_summary.get('Expense', 0)
)

monthly_summary
```

```python
category_expense = df[df['Income/Expense'] == 'Expense'].groupby('Category')['Amount'].sum().sort_values(ascending=False)

category_expense
```

```python
normal_expenses = df[
    (df['Income/Expense'] == 'Expense') &
    (~df['Category'].isin([
        'Money transfer',
        'Investment',
        'Public Provident Fund',
        'Recurring Deposit'
    ]))
]

normal_category_expense = (
    normal_expenses.groupby('Category')['Amount']
    .sum()
    .sort_values(ascending=False)
)

normal_category_expense
```

```python
# Fill missing values
df['Subcategory'] = df['Subcategory'].fillna('Unknown')
df['Note'] = df['Note'].fillna('No Note')

# Remove extra spaces from text columns
df['Mode'] = df['Mode'].str.strip()
df['Category'] = df['Category'].str.strip()
df['Income/Expense'] = df['Income/Expense'].str.strip()
df['Currency'] = df['Currency'].str.strip()

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Create separate amount columns for easier Power BI analysis

df['Expense_Amount'] = df['Amount'].where(
    df['Income/Expense'] == 'Expense', 0
)

df['Income_Amount'] = df['Amount'].where(
    df['Income/Expense'] == 'Income', 0
)

df['Transfer_Amount'] = df['Amount'].where(
    df['Income/Expense'] == 'Transfer-Out', 0
)

# Create Month-Year field
df['Month_Year'] = df['Date'].dt.to_period('M').astype(str)

# Create a proper date field for monthly sorting
df['Month_Year_Date'] = df['Date'].dt.to_period('M').dt.to_timestamp()

df.head()
```

```python
# Define essential and discretionary expense categories

essential_categories = ['Transportation', 'Household', 'Food', 'Family', 'Health', 'maid', 'Cook', 'Rent', 'water (jar /tanker)', 'Documents', 'garbage disposal', 'Education', 'Self-development']

discretionary_categories = ['subscription', 'Tourism', 'Gift', 'Apparel', 'Festivals', 'Culture', 'Beauty', 'Grooming', 'Social Life']

# Create spending type
df['Spending_Type'] = 'Other'

df.loc[(df['Income/Expense'] == 'Expense') & (df['Category'].isin(essential_categories)), 'Spending_Type'] = 'Essential'
df.loc[(df['Income/Expense'] == 'Expense') & (df['Category'].isin(discretionary_categories)), 'Spending_Type'] = 'Discretionary'

df[['Category', 'Spending_Type']].drop_duplicates().sort_values('Category')
```

```python
# Detect unusually high regular expenses using IQR

expense_mask = (df['Income/Expense'] == 'Expense') & (~df['Category'].isin(['Money transfer', 'Investment', 'Public Provident Fund', 'Recurring Deposit']))

Q1 = df.loc[expense_mask, 'Amount'].quantile(0.25)
Q3 = df.loc[expense_mask, 'Amount'].quantile(0.75)
IQR = Q3 - Q1
upper_limit = Q3 + (1.5 * IQR)

df['Unusual_Expense'] = 'No'
df.loc[expense_mask & (df['Amount'] > upper_limit), 'Unusual_Expense'] = 'Yes'

print("Q1:", Q1)
print("Q3:", Q3)
print("Upper Limit:", upper_limit)
print("Unusual Expenses:", (df['Unusual_Expense'] == 'Yes').sum())
```

```python
unusual_expenses = df[df['Unusual_Expense'] == 'Yes'][['Date', 'Category', 'Subcategory', 'Note', 'Amount', 'Spending_Type']]
unusual_expenses.sort_values('Amount', ascending=False)
```

```python
print("Unusual expense count by category:")
print(df[df['Unusual_Expense'] == 'Yes']['Category'].value_counts())

print("\nTotal amount of unusual expenses: ₹", round(df.loc[df['Unusual_Expense'] == 'Yes', 'Amount'].sum(), 2))
```

```python
unusual_summary = df[df['Unusual_Expense'] == 'Yes'].groupby('Category')['Amount'].agg(['count', 'sum']).sort_values('sum', ascending=False)

unusual_summary
```
